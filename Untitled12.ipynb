```python
# Full comparison: CNN, LSTM, CNN-LSTM, GRU (with evaluation + 2x2 figure)
import numpy as np
import pandas as pd
import yfinance as yf
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, GRU, Conv1D, MaxPooling1D, Flatten, Dropout

# --- CONFIG ---
SYMBOL = "GOOG"
START = "2015-01-01"
END   = "2024-01-01"
TIME_STEPS = 60
EPOCHS = 20
BATCH = 32
# -------------


# 1) LOAD DATA
df = yf.download(SYMBOL, start=START, end=END)
if df.empty:
    raise RuntimeError("No data downloaded — check SYMBOL/Internet.")
prices = df[['Close']].values.astype(float)

# 2) SCALE
scaler = MinMaxScaler(feature_range=(0,1))
scaled = scaler.fit_transform(prices)

# 3) CREATE SEQUENCES
def create_sequences(data, steps=60):
    X, y = [], []
    for i in range(steps, len(data)):
        X.append(data[i-steps:i, 0])
        y.append(data[i, 0])
    X = np.array(X)
    y = np.array(y)
    return X.reshape((X.shape[0], X.shape[1], 1)), y

X, y = create_sequences(scaled, TIME_STEPS)

# 4) SPLIT (80/20)
split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# Utility to inverse-transform predictions
def inv_transform(arr):
    return scaler.inverse_transform(arr.reshape(-1,1)).reshape(-1)

# 5) BUILD MODELS
def build_lstm(input_shape):
    m = Sequential([
        LSTM(50, return_sequences=True, input_shape=input_shape),
        LSTM(50),
        Dense(1)
    ])
    m.compile(optimizer='adam', loss='mse')
    return m

def build_gru(input_shape):
    m = Sequential([
        GRU(50, return_sequences=True, input_shape=input_shape),
        GRU(50),
        Dense(1)
    ])
    m.compile(optimizer='adam', loss='mse')
    return m

def build_cnn(input_shape):
    m = Sequential([
        Conv1D(64, kernel_size=3, activation='relu', input_shape=input_shape),
        MaxPooling1D(pool_size=2),
        Flatten(),
        Dense(50, activation='relu'),
        Dense(1)
    ])
    m.compile(optimizer='adam', loss='mse')
    return m

def build_cnn_lstm(input_shape):
    m = Sequential([
        Conv1D(64, kernel_size=3, activation='relu', input_shape=input_shape),
        MaxPooling1D(pool_size=2),
        LSTM(50),
        Dense(1)
    ])
    m.compile(optimizer='adam', loss='mse')
    return m

# 6) TRAIN & PREDICT helper
def train_and_predict(model, X_train, y_train, X_test, epochs=EPOCHS, batch=BATCH, verbose=1):
    model.fit(X_train, y_train, epochs=epochs, batch_size=batch, verbose=verbose)
    pred = model.predict(X_test)
    return pred.reshape(-1)

# 7) Run models (note: training will take time depending on hardware)
input_shape = (X_train.shape[1], X_train.shape[2])

print("Training LSTM...")
lstm_model = build_lstm(input_shape)
pred_lstm_scaled = train_and_predict(lstm_model, X_train, y_train, X_test)

print("Training CNN...")
cnn_model = build_cnn(input_shape)
pred_cnn_scaled = train_and_predict(cnn_model, X_train, y_train, X_test)

print("Training CNN-LSTM...")
cnn_lstm_model = build_cnn_lstm(input_shape)
pred_cnnlstm_scaled = train_and_predict(cnn_lstm_model, X_train, y_train, X_test)

print("Training GRU...")
gru_model = build_gru(input_shape)
pred_gru_scaled = train_and_predict(gru_model, X_train, y_train, X_test)

# 8) Inverse transform predicted & actual values
pred_lstm = inv_transform(np.array(pred_lstm_scaled))
pred_cnn = inv_transform(np.array(pred_cnn_scaled))
pred_cnnlstm = inv_transform(np.array(pred_cnnlstm_scaled))
pred_gru = inv_transform(np.array(pred_gru_scaled))

actual = inv_transform(y_test)

# 9) EVALUATE
def metrics(name, actual, predicted):
    mse = mean_squared_error(actual, predicted)
    rmse = np.sqrt(mse)
    r2 = r2_score(actual, predicted)
    print(f"\n{name} Performance:")
    print(f"  MSE : {mse:.6f}")
    print(f"  RMSE: {rmse:.6f}")
    print(f"  R2  : {r2:.6f}")
    return {'name': name, 'mse': mse, 'rmse': rmse, 'r2': r2}

res_lstm = metrics("LSTM", actual, pred_lstm)
res_cnn = metrics("CNN", actual, pred_cnn)
res_cnnlstm = metrics("CNN-LSTM", actual, pred_cnnlstm)
res_gru = metrics("GRU", actual, pred_gru)

# 10) SUMMARY: choose best by R2 (higher better) and by RMSE (lower better)
results = [res_lstm, res_cnn, res_cnnlstm, res_gru]
best_by_r2 = max(results, key=lambda x: x['r2'])
best_by_rmse = min(results, key=lambda x: x['rmse'])

print("\nBest by R² :", best_by_r2['name'], f"(R²={best_by_r2['r2']:.4f})")
print("Best by RMSE:", best_by_rmse['name'], f"(RMSE={best_by_rmse['rmse']:.4f})")

# 11) PLOT: 2x2 subplots Actual vs Predicted
plt.style.use('default')
fig, axes = plt.subplots(2, 2, figsize=(15,10))
axes = axes.flatten()

models_plot = [
    ("LSTM", actual, pred_lstm),
    ("CNN", actual, pred_cnn),
    ("CNN-LSTM", actual, pred_cnnlstm),
    ("GRU", actual, pred_gru),
]

for ax, (title, act, pred) in zip(axes, models_plot):
    ax.plot(act, label="Actual", linewidth=1.5)
    ax.plot(pred, label="Predicted", linestyle='dashed', linewidth=1.2)
    ax.set_title(f"{title} — Actual vs Predicted")
    ax.set_xlabel("Test samples (time order)")
    ax.set_ylabel("Price")
    ax.legend()
    ax.grid(True)

plt.tight_layout()
plt.show()

# 12) OPTIONAL: Save the figure
fig.savefig(f"{SYMBOL}_model_comparison.png", dpi=200)
print(f"\nSaved comparison figure as: {SYMBOL}_model_comparison.png")
```
